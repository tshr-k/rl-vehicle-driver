```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.interpolate import CubicSpline

plt.style.use("ggplot")
```

```python
g = 9.81
max_thrust = 11_000.0
max_brake_force = -38_000.0

throttle = np.array([-1, 0, 1])
driving_force = np.array([max_brake_force, 0, max_thrust])

cubic_spline_interpolator = CubicSpline(throttle, driving_force, bc_type=("natural", "clamped"))  # type: ignore
```

```python
x = np.linspace(-1, 1, 100)

# plt.plot(x, pchip_interpolator(x), label='Interpolated acceleration')
plt.plot(x, cubic_spline_interpolator(x), label="Cubic Spline Interpolation")
plt.scatter(throttle, driving_force, color="red", label="Waypoints")

plt.title("Throttle to Driving Force Mapping")
plt.xlabel("Throttle")
plt.ylabel("Driving Force (N)")
```
